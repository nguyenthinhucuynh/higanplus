## 7. View FID / KID / loss curves (live, while training runs)

`train.py` writes a TensorBoard event file to `runs/gan_iam_masking-<m-d-H-M>/`.  Run the cell below in a *second browser tab* (so the Train cell above keeps streaming logs) to mount that directory under TensorBoard, which Kaggle renders inline.

Scalars to check:

| Tag | Where it lives | Lower or higher better | What it measures |
| --- | --- | --- | --- |
| `valid/fid` | per epoch | **lower** | Frechet distance between Inception features of generated vs real images.  Below ~30 means the model's output distribution is close to IAM. |
| `valid/kid` | per epoch | **lower** | Kernel Inception Distance.  Less biased than FID at small validation sample counts. |
| `valid/is_gen` | per epoch | higher | Inception Score on generated images.  Diversity proxy. |
| `loss/mask_consistency_loss` | per `print_iter_val` | lower | **NEW**: Consistency loss for masking heads. Ensures vertical and horizontal masked styles are coherent with base style. |
| `loss/fake_ctc_loss` | per `print_iter_val` | lower | CTC loss the recognizer reports on fake images. |
| `loss/adv_loss` | per `print_iter_val` | track stability | G's adversarial loss vs D. |
| `loss/recn_loss` | per `print_iter_val` | lower | L1 reconstruction of `recn_imgs` against `real_imgs`. |

## 8. Resume helper (scoped to this experiment only)

Looks for the latest `runs/gan_iam_masking-*` directory and patches `configs/gan_iam_masking.yml` so the next `train.py` call starts from that run's `last.pth`.

```python
import glob, os, re

RUN_PREFIX = 'gan_iam_masking'
CFG_PATH = f'/kaggle/working/higanplus/HiGAN+/configs/gan_iam_masking.yml'

candidates = sorted(
    glob.glob(f'/kaggle/working/higanplus/HiGAN+/runs/{RUN_PREFIX}-*'),
    key=os.path.getmtime,
)
if not candidates:
    print(f'no {RUN_PREFIX} runs found yet -- nothing to resume from')
else:
    run_dir = candidates[-1]
    last_ckpt = os.path.join(run_dir, 'ckpts', 'last.pth')
    print('latest run :', run_dir)
    print('latest ckpt:', last_ckpt, '(exists:', os.path.exists(last_ckpt), ')')
    with open(CFG_PATH) as f: txt = f.read()
    new_txt = re.sub(r"pretrained_ckpt:\s*'[^']*'", f"pretrained_ckpt: '{last_ckpt}'", txt)
    if new_txt != txt:
        with open(CFG_PATH, 'w') as f: f.write(new_txt)
        print(f'patched {CFG_PATH} -> resume on next train.py call')
    else:
        print('config already references the latest ckpt; nothing to do')
```
